```python
import kagglehub
import pandas as pd
# Download latest version
path = kagglehub.dataset_download("mohitkumar282/used-car-dataset")
data = pd.read_csv(f"{path}/used_car_dataset.csv")
print("Path to dataset files:", path)
```

```python
data.dropna(inplace=True)
```

```python
import random

data['City']=random.choices(['Delhi', 'Mumbai', 'Bangalore', 'Chennai', 'Hyderabad'], k=len(data))
```

```python
data.shape
```

```python
data.head()
```

```python
data.drop(['Age'], axis=1, inplace=True)
```

```python
data.head()
```

```python
data.drop(['PostedDate'],axis=1,inplace=True)
```

```python
data['Description']=data['Brand'][0] + ' ' + data['model'][0] + ' ' + data['Year'].astype(str)[0] + ' model of ' + data['FuelType'][0] + ' variant driven for ' + data['kmDriven'].astype(str)[0] + ' in ' + data['City'][0]
```

```python
data.head()
```

```python
data.rename(columns={'Brand':'brand','model':'model','Year':'year','FuelType':'fuel_type','kmDriven':'km_driven','City':'city','AskPrice':'price','Description':'description','AdditionInfo':'additional_info','Transmission':'transmission'}, inplace=True)
```

```python
data['car_id']=random.sample(range(1000,100000),k=len(data))
```

```python
data.rename(columns={'Owner':'owner'},inplace=True)
```

```python
import os
from huggingface_hub import InferenceClient
from dotenv import load_dotenv
load_dotenv()
client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["hf_token"],
)
batch=32
for i in range(0, len(data), batch):
    batch_data = data['description'].iloc[i : i + batch].to_list()
    
    embeddings_list = client.feature_extraction(
        batch_data,
        model="microsoft/harrier-oss-v1-0.6b",
    )
    
    # 1. Convert the (32, 1024) array into a list of individual vectors
    embeddings_as_list = [v for v in embeddings_list]
    
    # 2. Get the specific indices for this batch
    batch_indices = data.index[i : i + len(embeddings_as_list)]
    
    # 3. Create a Series with the matching index to ensure perfect alignment
    data.loc[batch_indices, 'embeddings'] = pd.Series(embeddings_as_list, index=batch_indices)
        
```

```python
import os
import pandas as pd
from huggingface_hub import InferenceClient
from dotenv import load_dotenv

load_dotenv()

client = InferenceClient(
    provider="hf-inference",
    api_key=os.environ["hf_token"],
)

# Initialize the column as 'object' to store lists properly
data['embeddings'] = None
data['embeddings'] = data['embeddings'].astype(object)

batch = 32
for i in range(0, len(data), batch):
    batch_data = data['description'].iloc[i : i + batch].to_list()
    
    # 1. Fetch embeddings
    embeddings_list = client.feature_extraction(
        batch_data,
        model="microsoft/harrier-oss-v1-0.6b",
    )
    
    # 2. CRITICAL FIX: Convert NumPy array to a standard Python list of lists.
    # This ensures commas are present and prevents string truncation like "[2.8 ...]"
    if hasattr(embeddings_list, 'tolist'):
        embeddings_as_list = embeddings_list.tolist()
    else:
        embeddings_as_list = list(embeddings_list)
    
    # 3. Get indices and assign
    batch_indices = data.index[i : i + len(embeddings_as_list)]
    data.loc[batch_indices, 'embeddings'] = pd.Series(embeddings_as_list, index=batch_indices)

# 4. SAVE CORRECTLY: 
# Avoid .to_csv() as it will turn your vectors back into broken strings.
data.to_parquet('embeddings_results.parquet') 
print("Embeddings saved successfully to Parquet format.")
```
